This assignment focuses on the design, implementation, and enhancement of a Retrieval-Augmented Generation (RAG) system. The objective is to combine document retrieval with language models to produce accurate, context-aware, and reliable responses while minimizing hallucinations.

## Part 1: Basic RAG Pipeline

In the first part, a complete RAG system is developed by integrating the following components:

- Document loading using a transcript as the knowledge base  
- Text chunking to improve retrieval precision  
- Embeddings to represent text in vector form  
- A vector database (FAISS) for efficient similarity search  
- A retriever to fetch relevant document chunks  
- A language model to generate answers based on retrieved context  

This part demonstrates how grounding responses in retrieved data improves answer quality compared to standalone language models.

## Part 2: RAG-Fusion

The second part extends the basic RAG system by improving retrieval through RAG-Fusion.

Key enhancements include:

- Generating multiple query variations for a single question  
- Retrieving documents for each query  
- Combining ranked results using Reciprocal Rank Fusion (RRF)  
- Using the fused results to produce more complete answers  

This approach improves retrieval robustness and reduces dependence on a single query formulation.

## Part 3: Chaining for Extended Context

This section introduces advanced chaining techniques to enhance reasoning and flexibility.

This includes:

- Conditional routing to classify whether a query is transcript-related or general  
- Sequential chaining to expand grounded RAG answers with deeper explanation  
- A fallback mechanism using a general language model for unrelated queries  

These additions allow the system to handle a wider range of questions and provide more comprehensive responses.

## Step 1: Load the Movie Transcript PDF

First, we load the movie transcript PDF that will serve as the knowledge base for the RAG system.

```python
pdf_path = "movietranscript.pdf"  

loader = PyPDFLoader(pdf_path)
documents = loader.load()

print("Total pages loaded:", len(documents))
print("\nFirst page preview:\n")
print(documents[0].page_content[:1000])
print("\nMetadata of first page:", documents[0].metadata)
```

## Step 2: Split the Transcript into Chunks

A full PDF page can be too large for accurate retrieval. Therefore, the transcript is split into smaller chunks. This improves semantic search because the retriever can find more precise sections of the transcript.

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = text_splitter.split_documents(documents)

print("Total chunks created:", len(chunks))
print("\nSample chunk:\n")
print(chunks[0].page_content[:800])
print("\nSample chunk metadata:", chunks[0].metadata)
```

## Step 3: Generate Embeddings

Each chunk is converted into a numerical vector representation using a Hugging Face embedding model. These embeddings allow the system to compare user queries with transcript chunks based on semantic similarity.

```python
# embeddings = GoogleGenerativeAIEmbeddings(
#     model="gemini-embedding-2-preview"
# )
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

## Step 4: Store Embeddings in a Vector Database

The generated embeddings are stored inside a FAISS vector database. FAISS enables fast similarity search, which makes it suitable for retrieval tasks in RAG systems.

```python
vector_store = FAISS.from_documents(chunks, embeddings)

print("FAISS vector store created successfully.")
```

## Step 5: Create the Retriever

The retriever searches the vector database and returns the most relevant transcript chunks for a given query. In this notebook, we retrieve the top 4 most similar chunks.

```python
retriever = vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 4}
)
```

## 📌 Objective

In this section, the original RAG system is extended into a **RAG-Fusion** pipeline and evaluated on the **SQuAD dataset**.

The goal of RAG-Fusion is to improve retrieval quality by:
- generating multiple query variations for the same question
- retrieving document chunks for each variation
- combining the ranked results using **Reciprocal Rank Fusion (RRF)**
- generating the final answer from the fused context

Unlike Part 1, which uses a movie transcript, this part uses **SQuAD**, a benchmark question-answering dataset containing:
- context passages
- questions
- ground-truth answers

This makes it easier to compare Vanilla RAG and RAG-Fusion using answer-based evaluation metrics such as **F1 score**.

## 🧠 Why RAG-Fusion?

A standard RAG system retrieves documents using only the original user question.  
However, retrieval quality may suffer if the wording of the question does not closely match the wording in the source passage.

SQuAD is a useful dataset for this experiment because:
- it contains longer context passages
- questions may use wording different from the exact context
- multiple related details may be spread across the same passage

RAG-Fusion addresses this by generating multiple related search queries, retrieving results for each query, and then merging them into a stronger final context using Reciprocal Rank Fusion.

This helps the retriever become less dependent on a single phrasing of the question.

## ⚙️ High-Level Pipeline

The RAG-Fusion system works in the following steps:

1. The user asks a question  
2. The language model generates several related search queries  
3. The vector store retrieves chunks for each generated query  
4. Reciprocal Rank Fusion combines the retrieved results into a single ranked list  
5. The top fused chunks are passed to the LLM  
6. The LLM generates the final answer using the fused context

This allows the model to retrieve evidence from multiple perspectives rather than relying on a single search query.

## Generating the Dataset and Setting up the Vector Store

```python
from datasets import load_dataset
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS

dataset = load_dataset("squad", split="validation[:200]")

# Keep QA pairs
qa_pairs = [
    {
        "question": item["question"],
        "answer": item["answers"]["text"][0]
    }
    for item in dataset
]

# Deduplicate contexts first
unique_contexts = list(dict.fromkeys(item["context"] for item in dataset))

base_docs = [
    Document(page_content=context)
    for context in unique_contexts
]

# Chunk contexts so retrieval becomes harder and Fusion has a chance to help
fusion_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=60
)

fusion_docs = fusion_splitter.split_documents(base_docs)

fusion_vector_store = FAISS.from_documents(fusion_docs, embeddings)

print("Unique contexts:", len(unique_contexts))
print("Fusion chunks:", len(fusion_docs))
```

##  Step 1: Generate Multiple Search Queries

The first component of RAG-Fusion is a **query-generation chain**.

Instead of searching the vector database with only the original question, the model produces several related search queries that preserve the same meaning but use different wording.

This improves retrieval because different query variants may match different parts of the transcript.

```python
query_generation_prompt = PromptTemplate.from_template(
    """
You are helping a RAG system retrieve information from a question-answering passage dataset.

Generate 3 high-quality search queries that:
- preserve the original meaning of the question
- use different wording or phrasing
- help retrieve relevant passages from the dataset

Return only the queries, one per line, with no numbering.

Question:
{question}

Search Queries:
"""
)
```

```python
query_generation_chain = query_generation_prompt | llm | StrOutputParser()
```

```python
def generate_fusion_queries(question):
    raw_output = query_generation_chain.invoke({"question": question})

    generated_queries = []
    for line in raw_output.splitlines():
        line = line.strip()
        if not line:
            continue
        line = line.lstrip("-•1234567890. ").strip()
        if line:
            generated_queries.append(line)

    # Always include the original question as well
    all_queries = [question] + generated_queries

    # Remove duplicates while preserving order
    seen = set()
    unique_queries = []
    for q in all_queries:
        q_norm = q.lower().strip()
        if q_norm not in seen:
            seen.add(q_norm)
            unique_queries.append(q)

    return unique_queries
```

## Step 2: Retrieve Documents for Each Query

For each generated query, the vector store performs a similarity search and returns the most relevant transcript chunks.

At this stage, we do **not** answer the question yet.  
We first collect multiple ranked document lists so that they can later be fused together.

```python
def retrieve_documents_for_queries(queries, per_query_k=2):
    all_results = []

    for q in queries:
        docs = fusion_vector_store.similarity_search(q, k=per_query_k)
        all_results.append(docs)

    return all_results
```

## Step 3: Apply Reciprocal Rank Fusion (RRF)

Now we combine the retrieved ranked lists using **Reciprocal Rank Fusion**.

RRF gives each document a score based on how highly it appears across multiple ranked lists.  
If a document appears near the top of several retrieval results, it receives a higher fused score.

This is useful because:
- documents supported by multiple query perspectives rise to the top
- ranking becomes more stable
- retrieval is less dependent on one specific wording of the question

```python
def reciprocal_rank_fusion(results, rank_constant=40, final_k=6):
    fused_scores = defaultdict(float)
    doc_lookup = {}

    for docs in results:
        for rank, doc in enumerate(docs, start=1):
            key = (doc.metadata.get("page", "N/A"), doc.page_content)
            fused_scores[key] += 1.0 / (rank_constant + rank)
            doc_lookup[key] = doc

    reranked = sorted(fused_scores.items(), key=lambda x: x[1], reverse=True)

    fused_docs = [doc_lookup[key] for key, _ in reranked[:final_k]]
    fused_score_list = [(doc_lookup[key], score) for key, score in reranked[:final_k]]

    return fused_docs, fused_score_list
```

## Step 4: Build the Fusion Retriever

This function combines the earlier steps into a single retrieval pipeline:

- generate multiple queries
- retrieve documents for each query
- fuse the ranked results with RRF
- return the final fused document set

This creates the **retrieval side** of the RAG-Fusion system.

```python
def fusion_retrieve(question, per_query_k=4, final_k=6):
    queries = generate_fusion_queries(question)
    retrieval_results = retrieve_documents_for_queries(queries, per_query_k=per_query_k)
    fused_docs, fused_scores = reciprocal_rank_fusion(
        retrieval_results,
        rank_constant=60,
        final_k=final_k
    )

    return {
        "question": question,
        "queries": queries,
        "retrieval_results": retrieval_results,
        "fused_docs": fused_docs,
        "fused_scores": fused_scores
    }
```

## Step 5: Inspect the Generated Fusion Queries

Before generating final answers, it is useful to verify that the model is producing sensible query variants.

This step helps demonstrate that the system is exploring multiple ways of phrasing the same information need.

```python
test_question = qa_pairs[0]["question"]
fusion_debug = fusion_retrieve(test_question, per_query_k=2, final_k=4)

print("Original Question:\n", fusion_debug["question"])
print("\nGenerated Queries:")
for i, q in enumerate(fusion_debug["queries"], start=1):
    print(f"{i}. {q}")
```

## Step 6: Build the Final Answering Chain for RAG-Fusion

After the fused context has been created, it is passed into the language model using a prompt similar to the original RAG prompt.

The model is instructed to answer only from the retrieved transcript context.

```python
fusion_answer_prompt = PromptTemplate.from_template(
    """
Answer the question using only the context provided below.
If the answer is not present in the context, say exactly:
"I could not find the answer in the provided context."

Context:
{context}

Question:
{question}

Answer:
"""
)
```

```python
fusion_answer_chain = fusion_answer_prompt | llm | StrOutputParser()
```

```python
def answer_with_rag_fusion(question, per_query_k=4, final_k=6):
    retrieval_output = fusion_retrieve(question, per_query_k=per_query_k, final_k=final_k)

    context = "\n\n".join(doc.page_content for doc in retrieval_output["fused_docs"])

    answer = fusion_answer_chain.invoke({
        "context": context,
        "question": question
    })

    retrieval_output["answer"] = answer
    retrieval_output["context"] = context
    return retrieval_output
```

## Step 9: Evaluate RAG-Fusion Performance

In this step, the performance of the RAG-Fusion system is evaluated using the SQuAD dataset.

The evaluation includes:
- **F1 Score** to measure overlap between generated answers and ground-truth answers  
- **Answer Length** as a proxy for completeness  

This provides a simple quantitative measure of how well the RAG-Fusion system generates accurate and complete responses.

```python
evaluation_set = qa_pairs[:3]
```

```python
def compute_f1(pred, truth):
    pred_tokens = pred.lower().split()
    truth_tokens = truth.lower().split()

    common = set(pred_tokens) & set(truth_tokens)

    if len(common) == 0:
        return 0

    precision = len(common) / len(pred_tokens)
    recall = len(common) / len(truth_tokens)

    return 2 * (precision * recall) / (precision + recall)
```

```python
comparison_rows = []

for item in evaluation_set:
    q = item["question"]
    truth = item["answer"]

    fusion = answer_with_rag_fusion(q, per_query_k=2, final_k=4)["answer"]

    comparison_rows.append({
        "question": q,
        "ground_truth": truth,
        "fusion_answer": fusion,
        "fusion_f1": round(compute_f1(fusion, truth), 2),
        "fusion_length": len(fusion.split())
    })

comparison_df = pd.DataFrame(comparison_rows)
comparison_df
```

```python
avg_fusion_f1 = comparison_df["fusion_f1"].mean()

print("Average Fusion F1:", round(avg_fusion_f1, 2))
```

### 📊 How well does the F1 score reflect the quality of the generated answers?

**Answer:**  
The F1 score measures how much overlap exists between the generated answer and the ground truth, capturing both precision and recall. Higher scores indicate better alignment, but it may still penalize correct answers that use different wording. Therefore, it should be interpreted alongside qualitative evaluation.

**Answer:**

If the generated queries are poor, or if the transcript does not contain the answer at all, the final result may still be limited. It also introduces additional retrieval steps, which can increase latency.

## 📌 Objective
In this section, we extend the basic RAG system by introducing **chaining for extended context**. The goal is to allow the language model to go beyond direct retrieval and handle a wider range of questions more intelligently.

The extended system uses two forms of chaining:

1. **Sequential chaining**  
   The system first generates a grounded answer using the RAG pipeline, and then passes that answer into another chain for expansion and interpretation.

2. **Conditional routing**  
   The system first classifies the query.  
   - If the query is related to the movie transcript, it is sent through the **RAG + extension chain**.  
   - If the query is unrelated to the transcript, it is sent to a **general language model chain**.

This demonstrates how chaining can enable:
- multi-step reasoning
- broader context handling
- safer fallback behavior when retrieval is insufficient

## 🧠 Conceptual Overview

The chaining system in this part works as follows:

### Step 1: Classification
The user query is first analyzed by a classifier chain, which determines whether the question is related to the movie transcript or not.

### Step 2: Routing
Based on the classifier output:
- transcript-related questions are routed to the **RAG pipeline**
- general questions are routed to a **general LLM chain**

### Step 3: Extended Reasoning
If the question goes through the RAG pipeline, the answer is then passed into a second chain that expands the response with interpretation, explanation, and narrative significance.

This creates a more advanced pipeline than simple RAG, because the model can:
- answer transcript-based questions using retrieved evidence
- handle non-transcript questions without failing
- explain limitations when the answer is not found

## 🏗️ Architecture

The final system can be summarized as:

**User Question → Classifier Chain → Routing → (RAG Chain → Extension Chain) OR General Chain**

This means that the system first decides *where* the question should go, and then performs *multiple steps of reasoning* if transcript-based retrieval is needed.

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough, RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

## Step 1: Build a Classifier Chain

The classifier chain is used to determine whether a user question is related to the movie transcript or not.

It must return only one of two labels:

- `"movie transcript"`
- `"other"`

This classification step is important because it allows the system to decide whether retrieval is necessary.

```python
classifier_prompt = PromptTemplate.from_template(
    """
Classify the following question into exactly one of these two categories:

- movie transcript
- other

Only return one of the two labels.

Question:
{question}

Classification:
"""
)
```

```python
classifier_chain = (
    {"question": RunnablePassthrough()}
    | classifier_prompt
    | llm
    | StrOutputParser()
)
```

## Step 2: Build a General LLM Chain

A general LLM chain is used as a fallback for questions that are not related to the movie transcript.

This is useful because a RAG system should not always attempt retrieval if the question is outside the scope of the document. Instead, such questions can be answered directly by the language model.

```python
general_prompt = PromptTemplate.from_template(
    """
Answer the following question clearly and concisely.

Question:
{question}

Answer:
"""
)
```

```python
general_chain = (
    {"question": RunnablePassthrough()}
    | general_prompt
    | llm
    | StrOutputParser()
)
```

## Step 3: Build the Extension Chain

The extension chain is responsible for expanding the grounded answer produced by the RAG system.

This means:
- the first chain retrieves evidence and generates a grounded answer
- the second chain adds explanation, interpretation, and broader context

This enables the system to go beyond simply repeating retrieved text.

```python
extension_prompt = PromptTemplate.from_template(
    """
You are given:
1. A user question
2. A grounded answer from a RAG system

Your task is to expand the grounded answer in a thoughtful and clear way.

Rules:
- If the grounded answer says "I could not find the answer in the transcript", do not invent facts.
- In that case, explain that the answer could not be confirmed from the retrieved transcript chunks.
- Otherwise:
  1. Restate the grounded answer clearly
  2. Expand it with interpretation or explanation
  3. Explain why it matters in the story

Question:
{question}

Grounded Answer:
{grounded_answer}

Expanded Response:
"""
)
```

```python
extension_chain = extension_prompt | llm | StrOutputParser()
```

## Step 4: Create the Sequential Extended Context Chain

This function connects the basic RAG pipeline with the extension chain.

The process is:

1. Send the question to the RAG chain  
2. Get the grounded answer  
3. Pass the grounded answer into the extension chain  
4. Return both results  

This demonstrates **multi-step reasoning** because the output of one chain becomes the input to another.

```python
def extended_context_chain(question):
    rag_result = rag_chain.invoke({"input": question})
    grounded_answer = rag_result["answer"]

    expanded_answer = extension_chain.invoke({
        "question": question,
        "grounded_answer": grounded_answer
    })

    return {
        "question": question,
        "grounded_answer": grounded_answer,
        "expanded_answer": expanded_answer
    }
```

## Step 5: Add Routing Logic

Now we add conditional routing.

The routing logic works as follows:

- If the classifier predicts `"movie transcript"`, the question is sent through the **extended RAG pipeline**
- If the classifier predicts `"other"`, the question is sent to the **general chain**

This combines two types of chaining:
- **branching / routing**
- **sequential reasoning**

```python
def route_with_extension(info):
    classification = info["topic"].strip().lower()
    question = info["question"]

    if "movie transcript" in classification:
        result = extended_context_chain(question)
        return result["expanded_answer"]
    else:
        return general_chain.invoke(question)
```

```python
full_chain = (
    {
        "topic": classifier_chain,
        "question": RunnableLambda(lambda x: x["question"])
    }
    | RunnableLambda(route_with_extension)
)
```

## Step 6: Test the Full Chaining System

We now test the final system with different kinds of questions:

1. A question related to the movie transcript  
2. A question requiring extended interpretation  
3. A general question outside the transcript  

This demonstrates that the system can both:
- retrieve and reason over transcript content
- answer general questions through routing

```python
q1 = "Why is Cooper’s decision to leave his family important to the story?"
print("Question:", q1)
print("\nAnswer:\n", full_chain.invoke({"question": q1}))
```

```python
q2 = "What does the bookshelf scene suggest about the relationship between Cooper and Murph?"
print("Question:", q2)
print("\nAnswer:\n", full_chain.invoke({"question": q2}))
```

```python
q3 = "What is the largest planet in the solar system?"
print("Question:", q3)
print("\nAnswer:\n", full_chain.invoke({"question": q3}))
```

## Step 7: Test the “Answer Not Found” Case

An important part of this implementation is handling cases where the retriever cannot find enough evidence in the transcript.

In such cases, the RAG chain may return:

**"I could not find the answer in the transcript."**

The extension chain should not hallucinate. Instead, it should explain the limitation clearly and responsibly.

```python
q4 = "What does Cooper try to tell Murph in the bookshelf scene?"
result_q4 = extended_context_chain(q4)

print("Question:", result_q4["question"])
print("\nGrounded Answer:\n", result_q4["grounded_answer"])
print("\nExpanded Answer:\n", result_q4["expanded_answer"])
```
